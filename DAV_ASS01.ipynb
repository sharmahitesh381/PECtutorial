```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

```

# Generating a dataset with 1000 observations for the variable height:

```python
randomheights = random.normal(loc=6,scale=2,size = (1000))
df = pd.DataFrame(randomheights, columns=['Height'])
df.to_csv("randomheights.csv", index=False)
```

PART A

```python
# Mean
mean_value = np.mean(randomheights)

# Median
median_value = np.median(randomheights)

# Variance
variance_value = np.var(randomheights, ddof=1)

# Standard Deviation
std_dev_value = np.std(randomheights, ddof=1)

# Range (Max - Min)
range_value = np.ptp(randomheights)

# Print results
print(f"Mean: {mean_value}")
print(f"Median: {median_value}")
print(f"Variance: {variance_value}")
print(f"Standard Deviation: {std_dev_value}")
print(f"Range: {range_value}")
```

PART B

```python
plt.hist(randomheights, bins=30, density=True, edgecolor='pink')

x_values = np.linspace(min(randomheights), max(randomheights), 1000)  # Generate x values for smooth curve
pdf_values = norm.pdf(x_values, mean_value, std_dev_value)  # Compute the PDF

plt.plot(x_values, pdf_values, color='red', label="PDF (Normal Curve)")  # Plot PDF curve
plt.xlabel('Height')
plt.ylabel('Frequency')
plt.title('Histogram of Height')
plt.show()
```

 The histogram  follows a bell-shaped curve of the PDF, the data is likely normally distributed. Also the mean , median and mode are very close to each other
